```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

db_path = "../data/raw/names.sqlite"
conn = sqlite3.connect(db_path)
```

```python
def styled_table(df, caption, color="#1565C0"):
    """Zeigt ein DataFrame als formatierte Tabelle mit farbiger Caption an."""
    return df.style.set_caption(caption).set_table_styles([
        {
            "selector": "caption",
            "props": [
                ("background-color", color),
                ("color", "white"),
                ("font-size", "16px"),
                ("font-weight", "bold"),
                ("padding", "8px")
            ]
        }
    ])
```

```python
df_national = pd.read_sql_query("SELECT * FROM NationalNames", conn)
df_state = pd.read_sql_query("SELECT * FROM StateNames", conn)
```

What tables are there?

```python
query = """
SELECT name
FROM sqlite_master
WHERE type = 'table';
"""

tables_df = pd.read_sql_query(query, conn)
tables_df
```

```python
display(styled_table(df_national.sample(5), "Random Sample (5) - NationalNames", "#1565C0"))
```

```python
display(styled_table(df_state.sample(5), "Random Sample (5) - StateNames", "#2E7D32"))
```

```python
display(styled_table(df_national.head(), "First 5 Rows - NationalNames", "#1565C0"))
```

```python
display(styled_table(df_state.head(), "First 5 Rows - StateNames", "#2E7D32"))
```

```python
query = """
SELECT *
FROM NationalNames
ORDER BY Year DESC, Id DESC
LIMIT 5;
"""

last_five_df = pd.read_sql_query(query, conn)
display(styled_table(last_five_df, "Last 5 Rows of NationalNames"))
```

```python
query = """
SELECT Name, SUM(Count) AS TotalBirths
FROM NationalNames
WHERE Gender = 'M'
GROUP BY Name
ORDER BY TotalBirths DESC
LIMIT 10;
"""

top10_male = pd.read_sql_query(query, conn)
top10_male
```

```python
query = """
SELECT
    Name,
    SUM(Count) AS TotalBirths
FROM NationalNames
GROUP BY Name
ORDER BY TotalBirths DESC
LIMIT 5;
"""

top5_names_df = pd.read_sql_query(query, conn)
display(styled_table(top5_names_df, "Top 5 Most Popular National Names", "#F9A825"))
```

```python
# year range - national
query = """
SELECT
    MIN(Year) AS MinYear,
    MAX(Year) AS MaxYear
FROM NationalNames;
"""

df_years_national = pd.read_sql_query(query, conn)
display(styled_table(df_years_national, "Year Range - NationalNames", "#F9A825"))
```

```python
# year range - state
query = """
SELECT
    MIN(Year) AS MinYear,
    MAX(Year) AS MaxYear
FROM StateNames;
"""

df_years_state = pd.read_sql_query(query, conn)
display(styled_table(df_years_state, "Year Range - StateNames", "#F9A825"))
```

```python
# gender distribution
query = """
SELECT
    Gender,
    SUM(Count) AS Births
FROM NationalNames
GROUP BY Gender;
"""

df_gender = pd.read_sql_query(query, conn)
display(styled_table(df_gender, "Gender Distribution", "#F9A825"))
```

```python
query = """
SELECT
    Name,
    SUM(Count) AS TotalBirths
FROM NationalNames
GROUP BY Name
ORDER BY TotalBirths ASC
LIMIT 5;
"""

least_five_df = pd.read_sql_query(query, conn)
display(styled_table(least_five_df, "Top 5 Least Popular National Names", "#F9A825"))
```

```python
query = """
SELECT *
FROM (
    SELECT 'First Year' AS Period, Name, SUM(Count) AS Births
    FROM NationalNames
    WHERE Year = (SELECT MIN(Year) FROM NationalNames)
    GROUP BY Name
    ORDER BY Births DESC
    LIMIT 3
)
UNION ALL
SELECT *
FROM (
    SELECT 'Last Year' AS Period, Name, SUM(Count) AS Births
    FROM NationalNames
    WHERE Year = (SELECT MAX(Year) FROM NationalNames)
    GROUP BY Name
    ORDER BY Births DESC
    LIMIT 3
);
"""

top_names = pd.read_sql_query(query, conn)
display(top_names)
```
